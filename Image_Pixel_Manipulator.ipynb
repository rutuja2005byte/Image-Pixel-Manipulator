```python
import numpy as np;
import matplotlib.pyplot as plt;
```

```python
img = np.zeros((200,200,3), dtype=np.uint8)

img[10:60, 20:90] = np.full((50,70,3) , [255,0,0], dtype=np.uint8)
img[70:120, 20:90] = np.full((50, 70, 3), [0, 255, 0], dtype=np.uint8)
img[130:180, 20:90] = np.full((50, 70, 3), [0, 0, 255], dtype=np.uint8)

plt.imshow(img)
plt.title("Step 1: RGB Image")
plt.axis('off')
plt.show()
```

```python
img[10:20, 90:170, 0] = 255

img[70:150, 90:100, 1] = 255

img[160:170, 20:120, 2] = 255

plt.imshow(img)
plt.title("Step 2: Color Blocks and Stripes")
plt.axis("off")
plt.show()
```

```python
cropped = img[10:100, 20:120]

flipped = img[:, ::-1, :]

red_channel = img[:, :, 0]
green_channel = img[:, :, 1]
blue_channel = img[:, :, 2]

fig, axes = plt.subplots(2, 3, figsize=(10, 6))

axes[0, 0].imshow(cropped)
axes[0, 0].set_title("Cropped Image")

axes[0, 1].imshow(flipped)
axes[0, 1].set_title("Flipped Image")

axes[0, 2].imshow(red_channel, cmap="gray", vmin=0, vmax=255)
axes[0, 2].set_title("Red Channel")

axes[1, 0].imshow(green_channel, cmap="gray", vmin=0, vmax=255)
axes[1, 0].set_title("Green Channel")

axes[1, 1].imshow(blue_channel, cmap="gray", vmin=0, vmax=255)
axes[1, 1].set_title("Blue Channel")

axes[1, 2].axis("off")

for ax in axes.flat:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
img_float = img.astype(np.float32)

bright_img = np.clip(img_float + 50, 0, 255).astype(np.uint8)
dark_img = np.clip(img_float - 50, 0, 255).astype(np.uint8)

bright_mask = np.all(img > 200, axis=2)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))

axes[0].imshow(bright_img)
axes[0].set_title("Brightened")

axes[1].imshow(dark_img)
axes[1].set_title("Darkened")

axes[2].imshow(bright_mask, cmap="gray")
axes[2].set_title("Bright Pixel Mask")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
img_uint8 = img

img_float64 = img.astype(np.float64)

print("Image shape:", img.shape)
print("uint8 memory:", img_uint8.nbytes, "bytes")
print("float64 memory:", img_float64.nbytes, "bytes")
print("Memory ratio:", img_float64.nbytes)

plt.imshow(img_uint8)
plt.title("Final Image - PixelForge")
plt.axis("off")
plt.show()
```
