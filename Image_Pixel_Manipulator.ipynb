```python

import numpy as np
import matplotlib.pyplot as plt
```

```python

img = np.zeros((200, 200, 3), dtype=np.uint8)

plt.imshow(img)
plt.axis("off")
plt.show()
```

```python

img[10:60, 20:90] = [255, 0, 0]
img[70:120, 20:90] = [0, 255, 0]
img[130:180, 20:90] = [0, 0, 255]

img[10:180, 110:130] = [255, 255, 0]
img[10:180, 140:160] = [255, 0, 255]

plt.imshow(img)
plt.axis("off")
plt.show()
```

```python

cropped = img[10:100, 20:100]

plt.imshow(cropped)
plt.axis("off")
plt.show()
```

```python

flipped = img[:, ::-1]

plt.imshow(flipped)
plt.axis("off")
plt.show()
```

```python

red = img[:, :, 0]
green = img[:, :, 1]
blue = img[:, :, 2]

plt.figure(figsize=(10, 3))

plt.subplot(1, 3, 1)
plt.imshow(red, cmap="gray")
plt.title("Red")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(green, cmap="gray")
plt.title("Green")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blue, cmap="gray")
plt.title("Blue")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python

bright = np.clip(img.astype(np.float32) + 50, 0, 255).astype(np.uint8)

plt.imshow(bright)
plt.axis("off")
plt.show()
```

```python

dark = np.clip(img.astype(np.float32) - 50, 0, 255).astype(np.uint8)

plt.imshow(dark)
plt.axis("off")
plt.show()
```

```python

mask = np.any(img > 200, axis=2)

plt.imshow(mask, cmap="gray")
plt.axis("off")
plt.show()
```

```python

img_uint8 = np.zeros((200, 200, 3), dtype=np.uint8)
img_float64 = np.zeros((200, 200, 3), dtype=np.float64)

print("uint8 memory:", img_uint8.nbytes, "bytes")
print("float64 memory:", img_float64.nbytes, "bytes")
```
